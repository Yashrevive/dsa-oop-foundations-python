#***Linear Search :***

# Q1) find the index of an element from a list and print it, if not found , print -1 ?

**Approch ->** There are various ways to solve this problem , but in this solution I'll try to solve it using *Linear search algorithm*  also known as *Brute Force Solution*.

In this approch we'll iterate over evry element of a list till we find the target , when found we'll print its index , if not found we'll print -1.

**Cases ->**

*   target in middle
*   target at last
*   target at start
*   target is absent
*   Empty Elements
*   Elements are of different data type






```python
def find_target(lists,target):

    for i in range(len(lists)):
        if type(i) == type(target):
            if lists[i] == target: # check if element equals to target
                return i
    return -1 # returns -1 if target not in the lists

lists = [1,5,7,85,7962,5,7,5]
target = 85
print(find_target(lists,target))
```

#***Binary Search ->***

#Q2)Find an element in a Sorted list and return its index using Binary Search, if not found return -1 ?

**Approch ->** There are various ways to solve this question but we would solve it using *Binary Search algorithm* .

In this we'll first compare middle element wiith target if its same , then return the middle element index.

If middle<half ; search in first half of the sorted list similarly.

If middle>half ; search in second half

Search till TARGET is found , if not found return -1.




**Cases ->**

* Element in middle
* Element at start
* Element at End
* Element not Found
* List is Empty
* Different data type









##Original code ->



```python
def binary_search(array,target:int):
    start = 0
    end = len(array)-1
    for i in range(end):
        mid = (start+end)//2 # index of middle element
        if array[mid] == target:
            return mid
        elif array[mid]<target:
            start = mid
        elif array[mid]>target:
            end = mid
    return -1 # Output if target not in array

array = sorted([85,74,9,5,7,95,26,985,854,0])
target = 7
print(binary_search(array,target))
```

##Flaws ->
• **The Infinite Stuck State (start = mid / end = mid)**: Because you don't exclude mid by adding or subtracting 1, your search window stops shrinking when start and end become adjacent. For example, if you look for 985 in your array, the code gets stuck in a loop looking at the same index until the loop iterations run out, incorrectly returning -1.

• **The Size 1 & 2 Crash**: Using for i in range(end) breaks completely for tiny arrays. If your array has 1 element, end = 0, the loop runs 0 times, and it skips the element entirely. If it has 2 elements, the loop runs exactly once, failing to check the second index if it needs to.

• **Rigid Loop Iterations**: A for loop forces the algorithm to run a fixed number of times based on the initial array length. In Binary Search, the loop must be dynamic—running only while the search window is active (while start <= end).


##Improvised Code ->

```python
def binary_search(array,target:int):

    start = 0
    end = len(array)-1

    while start <= end: #loop

        if array[start] == target:
            return start
        elif array[end] == target:# checking first and last element
            return end

        mid = (start+end)//2 #index of middle element

        if array[mid] == target:
            return mid
        elif array[mid] > target: #algorithm
            end = mid-1
        elif array[mid] < target:
            start = mid+1

    return -1 # target not in array


array = sorted([85,74,9,5,7,95,26,985,854,0])
target = 85
print(binary_search(array,target))
```

##Special Case ->

if an element has multiple occurance.

Eg; [2,4,6,6,6,6,7,9,9,9,9,9]

##Solution ->

we need to add a check, so that we can verify if the mid is the first occourance or not, if not then we need to shift it.

##code ->

```python
def binary_search(array,target:int):

    start = 0
    end = len(array)-1

    while start <= end:

        if array[start]==target:
            return start

        mid = (start+end)//2

        if array[mid]==target:
            if mid-1>=0 and array[mid-1]==target:
                mid = mid-1
                end = mid
            else:
                return mid
        elif array[mid]>target:
            end = mid-1
        else:
            start = mid+1

    return -1

array = sorted([1,2,65,74,5,74,6,9,49,61,2,85,2,58,5,5,55,23,7,7,7,8,9,0])
target = 9
print(binary_search(array,target))
```

##**Inefficiency ->**

In our earlier Solution, time complexity was O(N),

but it's very inefficient for large data sets.

##Solution ->

To solve this INEFFICIENCY, we introduce a new variable, "result" it'll record the index of previous occurance, and we'll run the algorithm again,

**If we find the target again** , then update the "result" and ignore the right half using (end=mid-1)

**if did not find the target** , ignore the left half using (start=mid+1) and let the algorithm run till end

##Code ->

```python
def binary_search(array,target):

    start = 0
    end = len(array)-1
    result = -1

    while start<=end:

        if array[start]==target:
            return start

        mid = (start+end)//2

        if array[mid]==target:
            result = mid
            end = mid-1
        elif array[mid]>target:
            end = mid-1
        else:
            start = mid+1

    return result


array = sorted([1,2,2,6,8,1,0,6,71,5,2,7,5,5,5,5,5,5,5,5,5,7,8,9,2,9,9,0])
target = 9
# print(array)
print(binary_search(array,target))
```
