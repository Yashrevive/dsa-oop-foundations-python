#Constructor , Self ->

* **_ _ init _ _( )** ; CONSTRUCTOR (It's a method which is automatically called when an object is created.)

* **self** ; Refers to current object. It's used in Every Method (Function).

```python
class Student :

    def __init__(self):
        self.name = input('Name ; ')                 #Constructor ->
        self.roll_num = int(input('Roll number ; ')) #it'll be called
        self.marks = float(input('Marks ; '))        #automatically as the
                                                     #function is called 1

    def print_details(self):
        print(f'Name ; {self.name}')
        print(f'Roll Number ; {self.roll_num}')
        print(f'Marks ; {self.marks}')

    def print_status(self):
        if self.roll_num >= 33.3:
            print('PASS')
        else:
            print('FAIL')


s1 = Student()     # self -> s1
s1.print_details()
s1.print_status()

s2 = Student()     # self -> s2
s2.print_details()
s2.print_status()
```

#Encapsulation ->
Binding of Data and Method together and *restricting direct access to data*.To restrict the direct access , the concept of ***Private Variable*** is introduced.

###Private Variable ->
It's denoted by (**_ _**) . Example ; __balance

If user try to ***access or modify*** private variable we'll get an Error ie ;
***<class_name> object has no attribute '__privatevariable' .***  

```python
class Bank:

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance #Private Variable

acc1 = Bank('Yash',1000)
print(acc1.__balance) # Error
```

#Getter and Setter ->
These are the Methods(Functions) which allowed controlled access to private variable

**Getter ->** Read Value

**Setter ->** Modify Value Safely



#Property ->
This attribute is used to convert a function(Method) to a variable / data.

We can make a setter for that property.

```python
class Bank:

    def __init__(self,name,balance):
        self.name = name
        self.__balance = balance

    #getter
    def getBalance(self):
        print(f'Balance -> {self.__balance}')
        return

    #setter
    def deposit(self,amount):
        self.__balance += amount
        print(f'Amount Deposited -> {amount} , Balance -> {self.__balance}')
        return

    @property
    def details(self):
        print(self.name)
        print(self.__balance)
        return

acc1 = Bank('Yash',10000)
acc1.getBalance()
acc1.deposit(500)
print(acc1.details)
```

#Abstraction ->

it's used to hide backend operations from the users and only showing essential informations.

* Its used to create a framework for a class which inherits abstract class

###Note ->
**If a class inherits *Abstract class*  but doesn't create *abstractmethod*  then Error will occour.**

###Error ->

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self):
        pass

class UPI(Payment): #Cause Error
    def payu(self,amount):
        print(f' Rs{amount} Transaction done')

p1 = UPI()
p1.payu(500)
```

###No Error ->

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self):
        pass

class UPI(Payment): #No Error
    def pay(self,amount):
        print(f' Rs{amount} Transaction done')

p1 = UPI()
p1.pay(500)
```

#Inheritance ->
Procees in which child class inherits all methods and properties(variable) from Parent Class.

```python
class Animal:
    def eat(self):
        print('Eating')

class Cat(Animal): #Inherits from Animal
    def walk(self):
        print('Running')

class Fish(Animal): #Inherits from Animal
    def swim(self):
        print('Swiming')

class Bird(Animal): #Inherits from Animal
    def fly(self):
        print('Flying')


cat = Cat()
cat.eat()
cat.walk()

fish = Fish()
fish.eat()
fish.swim()

bird = Bird()
bird.eat()
bird.fly()
```

#Single Inheritance ->
Inheritance in which Child class inherits from 1 parent class.

##super() Keyword ->
Its used to acces constructor(_ _ init _ _( )) from parent class by a Child Class.

Also, its used to access methods(function) from parent class,  as we can't access a method from parent class if the child class has a method(function) as same name as of parent class method which you want to access.

```python
class Company:
    def __init__(self,companyname):
        self.companyname = companyname
    def address(self):
        print('Bangalore')
    def getname(self):
        print(self.companyname)

class Employee(Company):
    def __init__(self,name,id,companyname):
        self.name = name
        self.id = id
        super().__init__(companyname) #Access __init__() of Company
    def work(self):
        print('SWE')
    def getname(self):
        super().getname() #Access getname() from Company
        print(self.name)

emp1 = Employee('Yash',1678,'Google')
emp1.work()
emp1.address()
print(emp1.name,emp1.id,emp1.companyname)
emp1.getname()
```

#Multiple Inheritance ->
Inheritance in which 1 child class inherits from multiple parent class.

```python
class Father:
    def earn(self):
        print('Father Earns')

class Mother:
    def cook(self):
        print('Mother Cooks')

class Child(Father,Mother):
    def study(self):
        print('child Studies')

child = Child()
child.study() #Method of Child class
child.earn() #Method from Father
child.cook() #Method from Mother
```

#MultiLevel Inheritance ->
Inheritance in which 2 class inherits from 1st class and 3rd class inherits from 2nd class.

=> 3rd class inherits from both 1st and 2nd cuz 2nd inherits from 1st.

```python
class Father:
    def earn(self):
        print('Father Earns')

class Mother(Father):
    def cook(self):
        print('Mother Cooks')

class Child(Mother):
    def study(self):
        print('child Studies')

child = Child()
child.study() #Method of Child class
child.earn() #Method from Father via Mother
child.cook() #Method from Mother
```

#Hierarchical Inheritance ->
Multiple child class inherits from single parent class.

```python
class Animal:
    def eat(self):
        print('Eats')

class Dog(Animal):
    def bark(self):
        print('Barks')

class Cat(Animal):
    def meow(self):
        print('Meows')

dog = Dog()
dog.eat() #Method from Animal
dog.bark()

cat = Cat()
cat.eat() #Method from Animal
cat.meow()
```

#MRO (Method Resolution Order) ->
Order in which Python searches for a method when a method is called.

* Its used when a **child class** inherits from **multiple parent class** and more than 1 parent class have same method (Function).

* Python uses **C3 Linearization Algorithm**

###Rules ->

* Left to Right Order

* No parent is skipped

* Consistent Hierarchy

```python
class Father:
    def drive(self):
        print('Father Drives')

class Mother:
    def drive(self):
        print('Mother Drives')

class Child_1(Father,Mother): # Father is priortize
    def drive(self):
        super().drive()
        print('Child Drives')

class Child_2(Mother,Father): # Mother priortize
    def drive(self):
        super().drive()
        print('Child Drives')

c1 = Child_1()
c1.drive()
c2 = Child_2()
c2.drive()
```

#Polymorphism ->
Process by which same method behaves differently in different situations.

## Types ->

#Method(Function) Overiding ->

We use it when we want to use a method from parent class according to our condition.

```python
class Animal:
    def sound(self):
        print('Animal Sound')

class Dog(Animal): # We modified it cuz we need dog to bark
    def sound(self):
        print('Bark')

dog = Dog()
dog.sound()
```

#Operator Overiding ->
It's used to make operator behave differentky from how they behave normally.

```python
class Coordinate:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self,other):
        return Coordinate(self.x+other.x,self.y+other.y)

c1 = Coordinate(3,4)
c2 = Coordinate(4,5)
c3 = c1+c2

print(c3.x,c3.y)
```

##Note ->
We can also modify Builtin functions

```python
class Coordinate:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __str__(self):
        return f'X:{self.x} , Y:{self.y}'

c1 = Coordinate(3,4)
c2 = Coordinate(4,5)

print(c1) #print(c1) = print(str(c1))
print(c2) #print(c2) = print(str(c2))
```
